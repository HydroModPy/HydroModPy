```python
from IPython.core.debugger import set_trace as st
import matplotlib.pyplot as plt
#Extract surfex dat
import sys
sys.path.append('D:/SURFEX/')

sim = ['ACC1', 'BCC1', 'BNU1', 'CAN1', 'CAN2', 'CAN3', 'CAN4', 'CAN5', 'CNR1', 'CSI1', 'IPS1', 'MIR1', 'MIR2', 'MIR3', 'NOR1']
#sim = ['ACC1']
sce = ['historic', 'RCP2.6', 'RCP4.5', 'RCP6.0', 'RCP8.5']
var = ['REC']
date = [2030, 2050, 2100]
shp_file = 'C:/Users/alexa/Dropbox/HydroModPy/ouve/gis/watershed.shp'
surfex_file = 'D:/SURFEX/OUT/OUVE_surfex_data.h5'
KR_file = "C:/Users/alexa/Dropbox/HydroModPy/ouve/gis/K_R.txt"
dem = 'C:/Users/alexa/Dropbox/HydroModPy/ouve/gis/watershed_buff_dem.tif'
out_path="C:/Users/alexa/Dropbox/HydroModPy/"
KR = np.loadtxt(KR_file)

for i in sim:
    for j in sce:
        try:
            if j == 'historic':
                sim_id = i + '_' + j
                recharge = clim.surfex(surfex_file, sim=i, var='REC', sce=j).period_data.mean()
                K = KR*(recharge/1000)
                st()
                mod.modflow_model(dem_path=dem,
                              watershed='ouve', climatic=[recharge/1000], lay_number=1, 
                              thick=20, hyd_cond=K, porosity=0.1, coastal_aquifer =True,
                              model_name=sim_id,
                              model_folder=out_path)
                ext.extract_modflow(dem_path=dem,
                                watershed='ouve',
                                model_name=sim_id,
                                model_folder=out_path)
                print(sim_id, recharge)
            else:
                for d in date:
                    if d == 2030 and j == 'RCP2.6':
                        SLR = 0.15
                    if d == 2030 and j == 'RCP4.5':
                        SLR = 0.2
                    if d == 2030 and j == 'RCP8.5':
                        SLR = 0.2
                    if d == 2050 and j == 'RCP2.6':
                        SLR = 0.3
                    if d == 2050 and j == 'RCP4.5':
                        SLR = 0.3
                    if d == 2050 and j == 'RCP8.5':
                        SLR = 0.4
                    if d == 2100 and j == 'RCP2.6':
                        SLR = 0.6
                    if d == 2100 and j == 'RCP4.5':
                        SLR = 0.7
                    if d == 2100 and j == 'RCP8.5':
                        SLR = 1.1
                    sim_id = i + '_' + j + '_' + str(d)
                    start_date= str(d-1)+'-09-01'
                    end_date= str(d)+'-04-01'
                    recharge = clim.surfex(surfex_file, sim=i, var='REC', sce=j, start_date=start_date, end_date=end_date).period_data.mean()
                    mod.modflow_model(dem_path=dem,
                              watershed='ouve', climatic=[recharge/1000], lay_number=1, 
                              thick=20, hyd_cond=K, porosity=0.1, coastal_aquifer =True, SLR=SLR,
                              model_name=sim_id,
                              model_folder=out_path)
                    ext.extract_modflow(dem_path=dem,
                                watershed='ouve',
                                model_name=sim_id,
                                model_folder=out_path)
                    print(sim_id, recharge, SLR)
        except:
            pass
                
            
```

```python
from IPython.core.debugger import set_trace as st
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import ticker, cm, rc
rc('font',**{'family':'Palatino Linotype','sans-serif':['Computer Modern']})
rc('text', usetex=True)
font = {'size'   : 30, 'weight' : 'bold'}
rc('xtick', labelsize=30)
rc('ytick', labelsize=30)
rc('font', **font)
from osgeo import gdal, osr
sim = ['ACC1', 'BCC1', 'BNU1', 'CAN1', 'CAN2', 'CAN3', 'CAN4', 'CAN5', 'CNR1', 'CSI1', 'IPS1', 'MIR1', 'MIR2', 'MIR3', 'NOR1']
#sim = ['ACC1']
sce = ['historic', 'RCP2.6', 'RCP4.5', 'RCP8.5']
var = ['REC']
date = [2030, 2050, 2100]
dem = 'C:/Users/alexa/Dropbox/HydroModPy/ouve/gis/watershed_buff_dem.tif'
path = 'C:/Users/alexa/Dropbox/HydroModPy/ouve/'
geol = gdal.Open('C:/Users/alexa/Dropbox/HydroModPy/ouve/gis/GeoStructure.tif')
dem_geo = geol.GetRasterBand(1).ReadAsArray()

f= plt.figure(figsize=(20,10))
ax = f.add_subplot(111)
x = np.array([1,2,3,4])

compt=0
for i in sim:
    for j in sce:
        try:
            if j == 'historic':
                x = np.array([1])
                sim_id = i + '_' + j
                wtd = 'C:/Users/alexa/Dropbox/HydroModPy/ouve/' + sim_id + '/watertable_depth.tif'
                dem_wtd = gdal.Open(wtd)
                dem_data = dem_wtd.GetRasterBand(1).ReadAsArray()
                dem_data[(dem_geo!= 1) & (dem_geo!= 21)] = np.nan
                area = np.sum(~np.isnan(dem_data))
                flooded = np.nansum(dem_data<0)/area
                unsat = np.nansum((dem_data>0) & (dem_data<0.5))/area
                risk_sum = flooded + unsat
                ax.plot(x, flooded, 'o' ,c='b', Label ='Flooded', ms = 10)
                ax.plot(x, unsat, '^' ,c='g', Label =r'Watertable depth $<$ 0.5m', ms = 10)
                ax.plot(x, risk_sum, 's' ,c='r', Label ='Sum', ms = 10)
                if compt == 0:
                    plt.legend()
                    compt += 1
            else:
                for d in date:
                    if d == 2030 and j == 'RCP2.6':
                        x = np.array([3])
                        c = 'r'
                    if d == 2030 and j == 'RCP4.5':
                        c = 'r'
                        x = np.array([4])
                    if d == 2030 and j == 'RCP8.5':
                        c = 'r'
                        x = np.array([5])
                    if d == 2050 and j == 'RCP2.6':
                        c = 'b'
                        x = np.array([7])
                    if d == 2050 and j == 'RCP4.5':
                        m = 'o'
                        c = 'b'
                        x = np.array([8])
                    if d == 2050 and j == 'RCP8.5':
                        c = 'b'
                        x = np.array([9])
                    if d == 2100 and j == 'RCP2.6':
                        c = 'g'
                        x = np.array([11])
                    if d == 2100 and j == 'RCP4.5':
                        c = 'g'
                        x = np.array([12])
                    if d == 2100 and j == 'RCP8.5':
                        c = 'g'
                        x = np.array([13])
                    sim_id = i + '_' + j + '_' + str(d)
                    wtd = 'C:/Users/alexa/Dropbox/HydroModPy/ouve/' + sim_id + '/watertable_depth.tif'
                    dem_wtd = gdal.Open(wtd)
                    dem_data = dem_wtd.GetRasterBand(1).ReadAsArray()
                    dem_data[(dem_geo!= 1) & (dem_geo!= 21)] = np.nan
                    area = np.sum(~np.isnan(dem_data))
                    flooded = np.nansum(dem_data<0)/area
                    unsat = np.nansum((dem_data>0) & (dem_data<0.5))/area
                    risk_sum = flooded + unsat
                    ax.plot(x, flooded, 'o' ,c='b', ms = 10)
                    ax.plot(x, unsat, '^' ,c='g', ms = 10)
                    ax.plot(x, risk_sum, 's' ,c='r', ms = 10)
        except:
            pass
x = np.array([1,3,4,5,7,8,9,11,12,13])
my_xticks = ['Historic','RCP 2.6: 2030','RCP 2.6: 2050','RCP 2.6: 2100','RCP 4.5: 2030', 'RCP 4.5: 2050', 'RCP 4.5: 2100','RCP 8.5: 2030','RCP 8.5: 2050','RCP 8.5: 2100']
degrees = 70
plt.xticks(x, my_xticks, rotation=degrees)
plt.ylabel('Percent of study site affected')
plt.tight_layout()
                    
plt.savefig("C:/Users/alexa/Dropbox/HydroModPy/ouve/Figure/Figure.pdf", dpi=400, format='pdf')
plt.savefig("C:/Users/alexa/Dropbox/HydroModPy/ouve/Figure/Figure.png", dpi=400, facecolor='w',edgecolor='w', format='png', pad_inches=0.1)              
                    
                    
```
